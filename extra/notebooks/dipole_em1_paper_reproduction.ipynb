# Dipole EM1 Paper Reproduction

This notebook uses the package API to validate paper data and document the current compatibility-script command for EM1 training.

```python
from pathlib import Path
from smlr import PaperDipoleAdapter, validate_paper_dipole_data

repo = Path.cwd().resolve()
adapter = PaperDipoleAdapter(
    strength_dir=str(repo / 'dipoles_data_all' / 'total_strength'),
    observable_dir=str(repo / 'dipoles_data_all' / 'total_alphaD'),
)
report = validate_paper_dipole_data(adapter.strength_dir)
report.to_dict()
```

```python
import subprocess

cmd = [
    'uv', 'run', 'python', 'Dipole_polarizability/src/main_gpt2.py',
    '--strength-dir', 'dipoles_data_all/total_strength/',
    '--alphaD-dir', 'dipoles_data_all/total_alphaD/',
    '--filter-ranges', '{"p1":[0.4,1.8],"p2":[1.5,4.0]}',
    '--n', '10', '--retain', '0.5', '--fold', '2.0',
    '--ansatz', 'paper_dipole', '--width-model', 'affine',
    '--w-strength', '1.0', '--w-alphaD', '2.0', '--w-m1', '0.0',
    '--n-restarts', '1', '--seed0', '42', '--num-iter', '20000',
    '--print-every', '1000', '--plots', 'save', '--save-dir', 'Dipole_polarizability/runs_em1_notebook',
]
print(' '.join(cmd))
# subprocess.run(cmd, check=True)
```
